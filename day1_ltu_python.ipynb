```python
pricesum = 0

def ticketprice(age):
  global pricesum

  if age < 12:
    pricesum += 4
    price = "child price: $4"
  elif age >= 65:
    pricesum += 6
    price = "senior price: $6"
  else:
    pricesum += 8
    price = "regular price: $8"

  return price

while(True):
  try:
    age_input = int(input("enter the age fatty (enter -1 to finish): "))

    if age_input == -1:
      print(f"Your total is ${pricesum}")
      break

    print(ticketprice(age_input))
  except ValueError:
    print("enter valid number idiot")
    continue
```

```python
def grade_calculator(score):
  if score >= 100:
    return("Your grade is S")
  if score >= 90:
    return("Your grade is A")
  elif score >= 80:
    return("Your grade is B")
  elif score >= 70:
    return("Your grade is C")
  else:
    return("Your grade is F")

while(True):
  try:
    grade = int(input("enter user's score (-1 to exit): "))
    if grade == -1:
      break
    print(grade_calculator(grade))
  except ValueError:
    print("enter a number dumbass")
```

```python
wordlist = []
amount = int(input("how many words u wanna input "))

for i in range(amount):
  word = input("enter a word ")
  wordlist.append(word)

count = 0

for i in range(len(wordlist)):
  if len(wordlist[i]) > 5:
    count += 1

print(f"there are {count} words in this list longer than 5 letters")

```
